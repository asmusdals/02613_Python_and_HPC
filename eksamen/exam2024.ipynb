### Q4 og Q5

Multi-processing, da denne er mest native til python applicationer

```python
def process_number(n): 
    s=0
    for i in range(n):
        s += i
    return s
```

```python
print(process_number(4))
print(process_number(10))
```

Vi bør bruge dynamic, så slipper vi for ujævn fordeling af arbejdet.

```python
def abssting(x,y):
    return abs(x+y)

a = abssting(1,2)
b = abssting(a,-3)

c = abssting(2,-3)
d = abssting(c,1)


print(b,d)
```

Vi er at den ikke er associative, altså kan den ikke bruges på nuværende form fordi den tager abs undervejs. For at ændre det bør den bare lave almindelig sum, og så lave abs til sidst.
